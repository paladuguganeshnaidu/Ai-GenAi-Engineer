```python
import math
#step function
def step_function(x):
    if x>=0 :
        return True
    else:
        return False
z=-3
z1=3
print(step_function(z))
print(step_function(z1))
```

```python
values = [-5, -2, -1, 0, 1, 2, 5]

for x in values:
    output = step_function(x)
    print(f"x = {x:>2} → output = {output}")
```

```python
#sigmoid function
def sigmoid(x):
    return 1/(1+math.exp(-x))
values = [-5, -2, -1, 0, 1, 2, 5]

for x in values:
    output = sigmoid(x)
    print(f"x = {x:>2} → sigmoid = {output:.4f}")
```

```python
#Tanh
def tanh(x):
    return math.tanh(x)
for x in values:
    output = tanh(x)
    print(f"x = {x:>2} → tanh = {output:.4f}")
```

```python
#ReLU
def relu(x):
    return max(0, x)
for x in values:
    output = relu(x)
    print(f"x = {x:>2} → ReLU = {output}")
```

```python
#Leaky ReLU
def leaky_relu(x, alpha=0.01):
    if x > 0:
        return x
    else:
        return alpha * x
for x in values:
    output = leaky_relu(x)
    print(f"x = {x:>2} → Leaky ReLU = {output:.4f}")
```

```python
#softmax
def softmax(scores):
    exp_scores = []

    for score in scores:
        exp_scores.append(math.exp(score))

    total = sum(exp_scores)

    probabilities = []

    for value in exp_scores:
        probabilities.append(value / total)

    return probabilities
scores = [2.5, 1.0, 0.2]

probabilities = softmax(scores)

print(probabilities)
print("Sum:", sum(probabilities))
```
