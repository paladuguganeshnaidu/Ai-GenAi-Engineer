```python
import torch
from torch import nn
X = torch.tensor([
    [1, 5],
    [2, 6],
    [3, 6],
    [4, 7],
    [5, 7],
    [6, 8],
    [7, 8],
    [8, 9]
], dtype=torch.float32)
y = torch.tensor([
    0,
    0,
    0,
    1,
    1,
    1,
    1,
    1
], dtype=torch.float32)
model = nn.Linear(2, 1)
predictions = model(X)
loss_fn = nn.MSELoss()

loss = loss_fn(predictions, y)

print("Loss:", loss)
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
optimizer.zero_grad()
predictions = model(X)
loss = loss_fn(predictions, y)
loss.backward()
print("\n")
print(model.weight.grad)
print(model.bias.grad)
optimizer.step()
```

```python
import torch
from torch import nn

X = torch.tensor([
    [1, 5],
    [2, 6],
    [3, 6],
    [4, 7],
    [5, 7],
    [6, 8],
    [7, 8],
    [8, 9]
], dtype=torch.float32)

y = torch.tensor([
    [0],
    [0],
    [0],
    [1],
    [1],
    [1],
    [1],
    [1]
], dtype=torch.float32)

model = nn.Linear(2, 1)

loss_fn = nn.BCEWithLogitsLoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

print("Before:")
print("Weight:", model.weight)
print("Bias:", model.bias)
for epoch in range(1,100000):

    optimizer.zero_grad()

    predictions = model(X)

    loss = loss_fn(predictions, y)

    loss.backward()

    optimizer.step()

    if epoch % 10000 == 0:
        print(
            f"Epoch {epoch} | "
            f"Loss: {loss.item():.4f} | "
            f"Weight: {model.weight.data} | "
            f"Bias: {model.bias.data}"
        )
with torch.no_grad():
    logits = model(X)
    probabilities = torch.sigmoid(logits)
    predictions = (probabilities >= 0.5).float()

print("Probabilities:")
print(probabilities)

print("Predicted classes:")
print(predictions)

print("Actual:")
print(y)

accuracy = (predictions == y).float().mean()

print("Accuracy:", accuracy.item())
```
