```python
# First Learning Algorithm

# Dataset
X = [1, 2, 3, 4]
Y = [2, 4, 6, 8]

# Initial Weight
w = 0.5

# Learning Rate
lr = 0.01

# Training
for epoch in range(10):

    # -------------------------
    # 1. Calculate predictions
    # -------------------------

    total_loss = 0
    gradient = 0

    for x, y in zip(X, Y):

        prediction = w * x

        # Loss
        loss = (y - prediction) ** 2
        total_loss += loss

        # Gradient
        gradient += -2 * x * (y - prediction)

    # -------------------------
    # 2. Update weight
    # -------------------------

    w = w - lr * gradient

    # -------------------------
    # 3. Display results
    # -------------------------

    print(
        f"Epoch: {epoch + 1:2d} | "
        f"Weight: {w:.4f} | "
        f"Loss: {total_loss:.4f} | "
        f"Gradient: {gradient:.4f}"
    )
```

```python
import math

# Dataset
X = [
    [1, 1],
    [1, 2],
    [2, 1],
    [2, 2],
    [5, 5],
    [5, 6],
    [6, 5],
    [6, 6]
]

Y = [
    0,
    0,
    0,
    0,
    1,
    1,
    1,
    1
]

# Parameters
w1 = 0.1
w2 = 0.1
b = 0.0

lr = 0.1

# Initialize accumulators
total_loss = 0
total_gradient_w1 = 0
total_gradient_w2 = 0
total_gradient_b = 0

# Process every sample
for x, y in zip(X, Y):

    x1 = x[0]
    x2 = x[1]

    # Forward Pass
    z = (x1 * w1) + (x2 * w2) + b
    
    # Sigmoid
    prediction = 1 / (1 + math.exp(-z))

    # Loss
    loss = (y - prediction) ** 2

    total_loss += loss

    # -------------------------
    # Backpropagation
    # -------------------------

    # Loss -> Prediction
    dL_dp = -2 * (y - prediction)

    # Prediction -> z
    dp_dz = prediction * (1 - prediction)

    # Loss -> z
    dL_dz = dL_dp * dp_dz

    # Gradients
    gradient_w1 = dL_dz * x1
    gradient_w2 = dL_dz * x2
    gradient_b = dL_dz

    total_gradient_w1 += gradient_w1
    total_gradient_w2 += gradient_w2
    total_gradient_b += gradient_b

    print(
        f"x1={x1}, "
        f"x2={x2}, "
        f"z={z:.4f}, "
        f"prediction={prediction:.4f}, "
        f"loss={loss:.4f}, "
        f"grad_w1={gradient_w1:.4f}, "
        f"grad_w2={gradient_w2:.4f}, "
        f"grad_b={gradient_b:.4f}, "
        f"actual={y}"
    )


w1 = w1 - lr * total_gradient_w1
w2 = w2 - lr * total_gradient_w2
b = b - lr * total_gradient_b
print("\n==============================")
print("Training Step Complete")
print("==============================")

print(f"Total Loss       = {total_loss:.4f}")

print(f"Total Gradient w1 = {total_gradient_w1:.4f}")
print(f"Total Gradient w2 = {total_gradient_w2:.4f}")
print(f"Total Gradient b  = {total_gradient_b:.4f}")

print(f"\nUpdated w1 = {w1:.4f}")
print(f"Updated w2 = {w2:.4f}")
print(f"Updated b  = {b:.4f}")
```
