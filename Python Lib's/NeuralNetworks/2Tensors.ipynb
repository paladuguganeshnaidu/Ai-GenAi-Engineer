```python
import torch

# 1. Check GPU availability and set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Selected device:", device)

if torch.cuda.is_available():
    print("GPU Name:", torch.cuda.get_device_name(0))

    # 2. Create inputs, weights, and bias directly on the GPU
    X_gpu = torch.tensor([
        [1.0, 2.0],
        [3.0, 4.0]
    ], device=device)

    W_gpu = torch.tensor([
        [0.5],
        [0.8]
    ], device=device)

    b_gpu = torch.tensor([0.1], device=device)

    # 3. Perform matrix multiplication and addition on GPU
    z_gpu = X_gpu @ W_gpu + b_gpu

    # 4. Print GPU details
    print("\n--- GPU Output ---")
    print("z_gpu:\n", z_gpu)
    print("Device:", z_gpu.device)
    print("Shape:", z_gpu.shape)

    # 5. Cross-check against CPU
    X_cpu = X_gpu.cpu()
    W_cpu = W_gpu.cpu()
    b_cpu = b_gpu.cpu()
    z_cpu = X_cpu @ W_cpu + b_cpu

    print("\n--- Cross-Check ---")
    print("CPU result matches GPU result:", torch.allclose(z_cpu, z_gpu.cpu()))
else:
    print("CUDA is not available on this system. Please check your GPU setup or drivers.")
```

```python
import torch

x = torch.tensor(5.0)

w = torch.tensor(
    2.0,
    requires_grad=True
)

target = torch.tensor(20.0)

lr = 0.01
for epoch in range(12):

    # Forward
    prediction = x * w

    # Loss
    loss = (target - prediction) ** 2

    # Backward
    loss.backward()

    print(
        f"Epoch: {epoch + 1} | "
        f"Weight: {w.item():.4f} | "
        f"Prediction: {prediction.item():.4f} | "
        f"Loss: {loss.item():.4f} | "
        f"Gradient: {w.grad.item():.4f}"
    )

    # Update
    with torch.no_grad():
        w -= lr * w.grad

    # Clear gradient
    w.grad.zero_()

    print(f"Updated Weight: {w.item():.4f}")
    print()
```

```python
import torch
import torch.nn as nn


class SimpleModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.w = nn.Parameter(
            torch.tensor(2.0)
        )

    def forward(self, x):
        return x * self.w


model = SimpleModel()

x = torch.tensor(5.0)
target = torch.tensor(20.0)

prediction = model(x)

loss = (target - prediction) ** 2

loss.backward()

print("Prediction:", prediction.item())
print("Loss:", loss.item())
print("Weight:", model.w.item())
print("Gradient:", model.w.grad.item())
```

```python
import torch
import torch.nn as nn


# Dataset

X = torch.tensor([
    [1, 1],
    [1, 2],
    [2, 1],
    [2, 2],
    [5, 5],
    [5, 6],
    [6, 5],
    [6, 6]
], dtype=torch.float32)


Y = torch.tensor([
    [0],
    [0],
    [0],
    [0],
    [1],
    [1],
    [1],
    [1]
], dtype=torch.float32)


# Model

model = nn.Sequential(
    nn.Linear(2, 1),
    nn.Sigmoid()
)


# Loss function

loss_function = nn.MSELoss()


# Learning rate

lr = 0.1


# Training

for epoch in range(10000):

    # Forward
    prediction = model(X)

    # Loss
    loss = loss_function(
        prediction,
        Y
    )

    # Backward
    loss.backward()

    # Update parameters
    with torch.no_grad():

        for parameter in model.parameters():

            parameter -= lr * parameter.grad

    # Clear gradients
    for parameter in model.parameters():

        parameter.grad.zero_()

    if (epoch + 1) % 1000 == 0:

        print(
            f"Epoch: {epoch + 1:3d} | "
            f"Loss: {loss.item():.4f}"
        )
```

```python
import torch
import torch.nn as nn


# Dataset

X = torch.tensor([
    [1, 1],
    [1, 2],
    [2, 1],
    [2, 2],
    [5, 5],
    [5, 6],
    [6, 5],
    [6, 6]
], dtype=torch.float32)


Y = torch.tensor([
    [0],
    [0],
    [0],
    [0],
    [1],
    [1],
    [1],
    [1]
], dtype=torch.float32)


# Model

model = nn.Sequential(
    nn.Linear(2, 1),
    nn.Sigmoid()
)


# Loss function

loss_function = nn.MSELoss()


# Learning rate

lr = 0.1


# Training

for epoch in range(10000):

    # Forward
    prediction = model(X)

    # Loss
    loss = loss_function(
        prediction,
        Y
    )

    # Backward
    loss.backward()

    # Update parameters
    with torch.no_grad():

        for parameter in model.parameters():

            parameter -= lr * parameter.grad

    # Clear gradients
    for parameter in model.parameters():

        parameter.grad.zero_()

    if (epoch + 1) % 1000 == 0:

        print(
            f"Epoch: {epoch + 1:3d} | "
            f"Loss: {loss.item():.4f}"
        )
print("\nFinal Predictions:")

with torch.no_grad():
    predictions = model(X)

    for x, prediction, actual in zip(X, predictions, Y):
        print(
            f"Input: {x.tolist()} | "
            f"Prediction: {prediction.item():.4f} | "
            f"Actual: {actual.item():.0f}"
        )
```
